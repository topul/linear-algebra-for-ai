## 一、NumPy 常用操作

````python
import numpy as np

# ===== 创建向量/矩阵 =====

a = np.array([1, 2, 3])              # 从列表创建 1 维数组（向量）
b = np.array([[1, 2], [3, 4]])        # 从嵌套列表创建 2 维数组（矩阵）

print('向量 a:', a)
print('矩阵 b:')
print(b)
````

````python
# ===== .shape 看形状 =====
# 这是你最常用的属性，看看数组长什么样

print('a 的形状:', a.shape)   # (3,) → 一维，3个元素
print('b 的形状:', b.shape)   # (2, 2) → 二维，2行2列

# 记忆：(行数, 列数)
````

````python
# ===== 常见的快捷创建 =====

print('全0:', np.zeros(5))                # [0. 0. 0. 0. 0.]
print('全1:', np.ones(3))                 # [1. 1. 1.]
print('等差:', np.arange(0, 1, 0.2))      # [0.  0.2 0.4 0.6 0.8]
print('随机:', np.random.randn(4))        # 4个随机数（正态分布）
````

````python
# ===== 索引（取元素）=====

v = np.array([10, 20, 30, 40, 50])

print('第0个:', v[0])           # 10 （从0开始数！）
print('最后一个:', v[-1])       # 50
print('第1到3:', v[1:4])        # [20 30 40]（左闭右开）

# 矩阵取值
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print('第0行:', m[0])           # [1 2 3]
print('第1列:', m[:, 1])        # [2 5]  （: 表示所有行）
print('第0行第2列:', m[0, 2])   # 3
````

````python
# ===== 数学运算 =====

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print('逐元素加:', x + y)         # [5 7 9]
print('逐元素乘:', x * y)         # [4 10 18]（不是矩阵乘！）
print('标量乘:  ', 3 * x)         # [3 6 9]
print('点积:    ', np.dot(x, y))  # 1*4 + 2*5 + 3*6 = 32
print('点积(另一种写法):', x @ y)  # 跟 np.dot 一样
````

````python
# ===== np.linalg —— 线性代数工具箱 =====
# linalg = linear algebra 的缩写

v = np.array([3, 4])

# norm = 范数（向量长度）
print('向量长度:', np.linalg.norm(v))   # √(9+16) = 5

# 以后还会用到：
# np.linalg.inv(M)    → 逆矩阵
# np.linalg.eig(M)    → 特征值/特征向量
# np.linalg.svd(M)    → 奇异值分解
````

````python
# ===== np.sqrt / np.exp / np.log =====
# numpy 的数学函数可以直接作用于整个数组

arr = np.array([1, 4, 9, 16])
print('开根号:', np.sqrt(arr))     # [1. 2. 3. 4.]
print('平方:  ', arr ** 2)          # [1 16 81 256]
print('指数:  ', np.exp([0, 1]))    # [1.  2.718...]
````

````python
# ===== f-string 格式化（Python，不是numpy，但课里经常用）=====

value = 3.14159
print(f'保留3位小数: {value:.3f}')   # 3.142
print(f'百分比: {0.95:.1%}')         # 95.0%

# {变量:.3f} 的意思：冒号后面是格式说明
#   .3 = 保留3位小数
#   f  = 浮点数格式
````

## 二、Matplotlib 常用操作

matplotlib 是 Python 的画图库。核心概念只有两个：

| 概念 | 比喻 | 代码 |
|------|------|------|
| **Figure** | 整张画布（纸） | `fig = plt.figure()` |
| **Axes** | 画布上的一个坐标系（图表） | `ax = fig.add_subplot()` |

一张画布上可以有多个图表（子图）。

````python
import matplotlib.pyplot as plt
%matplotlib inline

# ===== 最简单的画图 =====

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 3, 5]

plt.plot(x, y)         # 画折线
plt.title('My Plot')   # 标题
plt.show()             # 显示
````

````python
# ===== plt.subplots —— 创建多个子图 =====
#
# fig, axes = plt.subplots(行数, 列数, figsize=(宽, 高))
#
# - fig  = 整张画布
# - axes = 所有子图的数组
#   如果是 (1, 2) 就有 axes[0] 和 axes[1] 两个图

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 左图
axes[0].plot([1, 2, 3], [1, 4, 9])
axes[0].set_title('Left')    # 单个子图用 set_title

# 右图
axes[1].bar([1, 2, 3], [3, 1, 2])
axes[1].set_title('Right')

plt.tight_layout()   # 自动调整间距，防止标题重叠
plt.show()
````

````python
# ===== ax.annotate —— 画箭头 =====
#
# 这就是课里画「向量箭头」的方法
#
# ax.annotate(文字,
#             xy=(箭头尖端x, 箭头尖端y),      ← 箭头指向哪里
#             xytext=(箭头起点x, 箭头起点y),   ← 箭头从哪里出发
#             arrowprops=dict(...)            ← 箭头的样式
#            )

fig, ax = plt.subplots(figsize=(5, 5))

# 从原点 (0,0) 画一个箭头到 (3, 4)
ax.annotate('',                         # 文字留空（不在箭头上写字）
            xy=(3, 4),                  # 箭头尖端（终点）
            xytext=(0, 0),              # 箭头起点
            arrowprops=dict(
                arrowstyle='->',        # 箭头样式
                lw=2,                   # lw = line width 线宽
                color='blue'
            ))

# 在箭头尖端旁边标注文字
ax.text(3.1, 4.1, '向量(3,4)', fontsize=12)

ax.set_xlim(-1, 5); ax.set_ylim(-1, 5)
ax.set_aspect('equal')   # x轴y轴刻度等长，不会变形
ax.grid(alpha=0.3)       # alpha=0.3 表示网格线半透明
ax.axhline(0, color='k', linewidth=0.5)  # 画x轴（水平线 y=0）
ax.axvline(0, color='k', linewidth=0.5)  # 画y轴（垂直线 x=0）
plt.show()
````

````python
# ===== ax.imshow —— 画热力图 =====
#
# 这就是课里画「相似度矩阵」的方法
# 把一个二维数组画成颜色方块图

data = np.array([[1.0, 0.9, 0.5],
                 [0.9, 1.0, 0.6],
                 [0.5, 0.6, 1.0]])

fig, ax = plt.subplots(figsize=(5, 4))

im = ax.imshow(data,             # 传入二维数组
               cmap='YlOrRd',    # cmap = color map 配色方案
                                 # YlOrRd = 黄→橙→红
               vmin=0.4,         # 颜色最浅对应的值
               vmax=1.0)         # 颜色最深对应的值

# 在每个格子中间写数字
labels = ['A', 'B', 'C']
ax.set_xticks(range(3))          # x轴刻度位置：0, 1, 2
ax.set_xticklabels(labels)       # 把 0,1,2 替换成 A,B,C
ax.set_yticks(range(3))
ax.set_yticklabels(labels)

for i in range(3):
    for j in range(3):
        # ax.text(x坐标, y坐标, 显示的文字, ...)
        # ha='center' = 水平居中, va='center' = 垂直居中
        ax.text(j, i, f'{data[i,j]:.2f}',
                ha='center', va='center', fontsize=14)

plt.colorbar(im, ax=ax)   # 在右边加一个颜色条（图例）
ax.set_title('Similarity Matrix')
plt.show()
````

````python
# ===== 常用设置速查 =====

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [1, 4, 2])

ax.set_xlim(0, 4)            # x轴范围
ax.set_ylim(0, 5)            # y轴范围
ax.set_xlabel('X axis')      # x轴标签
ax.set_ylabel('Y axis')      # y轴标签
ax.set_title('Title')        # 标题
ax.set_aspect('equal')       # 等比例（x和y同刻度）
ax.grid(True, alpha=0.3)     # 显示网格
ax.legend(['line 1'])        # 图例

plt.tight_layout()
plt.show()
````

## 三、课程里常见的「套路」总结

看课程代码时不要逐字读，抓住这几个模式就够了：

### 模式1：画向量箭头
```python
ax.annotate('', xy=终点, xytext=起点,
            arrowprops=dict(arrowstyle='->', lw=2, color='blue'))
ax.text(x, y, '标注文字')   # 在旁边写字
```

### 模式2：画热力图（相似度矩阵）
```python
im = ax.imshow(二维数组, cmap='YlOrRd', vmin=最小, vmax=最大)
ax.set_xticks(...)         # 设x轴标签
ax.set_yticks(...)         # 设y轴标签
for i, j:                  # 在每个格子写数字
    ax.text(j, i, ...)
plt.colorbar(im, ax=ax)    # 颜色条
```

### 模式3：计算再输出
```python
result = np.dot(a, b)                  # 某个 numpy 运算
print(f'描述文字: {result:.3f}')       # .3f = 保留3位小数
```
